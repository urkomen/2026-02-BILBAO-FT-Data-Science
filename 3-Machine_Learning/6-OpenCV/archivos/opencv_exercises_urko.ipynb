Ejercicio 2: Graba un video de unos segundos de tu cara moviéndose por la pantalla (que no se quede fija en el centro solamente) con tu webcam del ordenador. Haz un detector de caras con dicho vídeo. Asegúrate de que tu cara está correctamente iluminada, ya que el modelo es bastante sensible.

Extra: intenta hacerlo con vídeo en directo tomado desde tu webcam.

```python
# Cargar el clasificador
face_cascade = cv.CascadeClassifier(cv.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Abrir el vídeo (0 = webcam, o ruta a un archivo)
# cap = cv.VideoCapture(0)
video_path = './Resources/Exercise/rostro.mp4'
cap = cv.VideoCapture(video_path)

# Tamaño de la ventana
cv.namedWindow('Detector de caras', cv.WINDOW_NORMAL)
cv.resizeWindow('Detector de caras', 860, 540)  # ajusta el tamaño que quieras

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Convertir a gris (el clasificador trabaja en gris)
    gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)

    # Detectar caras
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,    # cuánto se reduce la imagen en cada escala
        minNeighbors=5,     # cuántos vecinos necesita para confirmar detección
        minSize=(30, 30)    # tamaño mínimo de cara a detectar
    )

    # Dibujar rectángulo en cada cara detectada
    for (x, y, w, h) in faces:
        cv.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv.putText(frame, 'Cara', (x, y-10),
                   cv.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)

    # Escalar solo para visualizar
    frame_small = cv.resize(frame, (0, 0), fx=0.2, fy=0.2)  # 20% del tamaño
    cv.imshow('Detector de caras', frame)

    # Pulsa Q para salir
    if cv.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv.destroyAllWindows()
```

Había una vez, un caracol con el que se metían mucho. Tanto era así, que decidió apuntarse a un gimnasio para ponerse súper fuerte para darle una paliza a todo el mundo.
```python
# Visualizamos componentes
snail = cv.imread('./Resources/Exercise/snail.jpg')
gym = cv.imread('./Resources/Exercise/gym.jpg')

snail_small = cv.resize(snail, None, fx=0.4, fy=0.4)
gym_small = cv.resize(gym, None, fx=0.2, fy=0.2)

cv.imshow('Caracol', snail_small)
cv.imshow('Gimnasio', gym_small)

cv.waitKey(0) # Necesitamos poner esta línea para indicarle que se cierre la ventana emergente manualmente sin que dé error en el notebook.
cv.destroyAllWindows()
```

Usamos la técnica de umbralización, debido a que tenemos fondo blanco y eso nos facilita muchísimo el trabajo de aislar el caracol

```python
# Calculamos máscara de caracol
img_rgb = cv.cvtColor(snail_small, cv.COLOR_BGR2RGB)

# Convertimos a gris
gray = cv.cvtColor(snail_small, cv.COLOR_BGR2GRAY)

# Umbralización inversa: fondo blanco → negro, caracol → blanco
_, mask = cv.threshold(gray, 240, 255, cv.THRESH_BINARY_INV)

# Cerramos pequeños agujeros en la máscara
kernel = np.ones((5, 5), np.uint8)
mask = cv.morphologyEx(mask, cv.MORPH_CLOSE, kernel)

# Aplicamos máscara
caracol_mask = cv.bitwise_and(img_rgb, img_rgb, mask=mask)

cv.imshow('Caracol', caracol_mask)

cv.waitKey(0)
cv.destroyAllWindows()
```

Metemos al caracol en el gimnasio

```python
# Cargar imagen de fondo y caracol
fondo_rgb = cv.cvtColor(cv.GaussianBlur(gym_small, (7,7), cv.BORDER_DEFAULT), cv.COLOR_BGR2RGB)
snail_rgb = cv.cvtColor(snail_small, cv.COLOR_BGR2RGB)

# Caracol pequeñito
escala = 0.5
nuevo_ancho = int(snail_rgb.shape[1] * escala)
nuevo_alto  = int(snail_rgb.shape[0] * escala)
snail_small2 = cv.resize(snail_rgb, (nuevo_ancho, nuevo_alto))

# Redimensionar la máscara que ya tenías calculada
mask_small      = cv.resize(mask, (nuevo_ancho, nuevo_alto))
mask_inv_small  = cv.bitwise_not(mask_small)

# Colocamos al caracol en las coordenadas x1, y1
x1, y1 = 250, 250  # cambia aquí tus coordenadas
region = fondo_rgb[y1:y1+nuevo_alto, x1:x1+nuevo_ancho]
caracol_layer = cv.bitwise_and(snail_small2, snail_small2, mask=mask_small)
fondo_layer   = cv.bitwise_and(region, region, mask=mask_inv_small)
fondo_rgb[y1:y1+nuevo_alto, x1:x1+nuevo_ancho] = cv.add(caracol_layer, fondo_layer)
```

Añadimos texto

```python
# Añadir texto encima de todo
cv.putText(fondo_rgb,
           '#gymsnail',
           (30, fondo_rgb.shape[0] - 20), # posición (x, y)
           cv.FONT_HERSHEY_SIMPLEX,
           2,                             # tamaño
           (0, 255, 0),                   # color blanco
           5,                             # grosor
           cv.LINE_AA)                    # antialiasing


# cv maneja bgr, convertimos a rgb antes de mostrar resultado
cv.imshow('Caracol en el gym', cv.cvtColor(fondo_rgb, cv.COLOR_RGB2BGR))

cv.waitKey(0)
cv.destroyAllWindows()
```
